# Quintessence analysis: fitting $(w_0, w_a)$ to thawing dark energy

Analysis of numerical products from `solver.py`, `fits.py`, `convergence.py`, and `diagnostics.py`. Reproduces the qualitative results of Shlivko & Steinhardt (2024) for two quintessence model families and characterizes the reliability of the $(w_0, w_a)$ compression.

The central question: when is
$$w_{\rm fit}(z) = w_0 + w_a \frac{z}{1+z}$$
a faithful description of the underlying scalar-field dynamics?

```python
import numpy as np
import matplotlib.pyplot as plt
import os, glob

plt.rcParams.update({
    'font.size': 11, 'figure.dpi': 110,
    'axes.grid': True, 'grid.alpha': 0.3,
    'axes.spines.top': False, 'axes.spines.right': False,
    'lines.linewidth': 1.5,
})

DATA = 'data/quintessence'
FITS = 'data/fits'
CONV = 'data/convergence'
DIAG = 'data/diagnostics'
FIGS = 'figures'
os.makedirs(FIGS, exist_ok=True)

N_MIN = -np.log(5.0)
N_FIT = np.linspace(N_MIN, 0.0, 401)
```

## 1. Quintessence dynamics

The scalar field $\phi(N)$ in $N = \ln a = -\ln(1+z)$ obeys
$$\phi'' = \tfrac{3}{2}\phi'(w_Q\Omega_Q - 1) - \frac{V_{,\phi}}{H_Q^2}, \qquad
\Omega_Q' = 3w_Q\Omega_Q(\Omega_Q - 1)$$
with the Friedmann constraint
$$H_Q^2 = \frac{V(\phi)}{3\Omega_Q - \tfrac{1}{2}\phi'^{\,2}}, \qquad
w_Q = \frac{\tfrac{1}{2}H_Q^2\phi'^{\,2} - V}{\tfrac{1}{2}H_Q^2\phi'^{\,2} + V}.$$

Initial conditions are set deep in matter domination with $\Omega_Q(N_i) = 10^{-6}$ and $\phi'(N_i) = 0$. We integrate forward in an internal variable $s$ until $\Omega_Q = 0.7$, then shift $N = s - s_f$ so that the present is at $N = 0$.

Two potentials:
$$V_{\rm exp}(\phi) = e^{\lambda\phi}, \qquad V_{\rm hill}(\phi) = 1 - \tfrac{1}{2}k^2\phi^2.$$

For the hilltop case we use $k^2 = 1$, which keeps $|\phi| < \sqrt{2}/k \approx 1.41$ and admits a robust scan over $\phi_i \in [0.001, 0.5]$.

```python
def load_sol(path):
    d = np.load(path)
    return dict(N=d['N'], phi=d['phi'], u=d['u'],
                Omega_Q=d['Omega_Q'], w_Q=d['w_Q'], H_Q=d['H_Q'])

# Representative solutions
reps = [
    ('exp,  $\lambda=0.5$',       f'{DATA}/exp_lam0.50/solution.npz'),
    ('exp,  $\lambda=1.0$',       f'{DATA}/exp_lam1.00/solution.npz'),
    ('exp,  $\lambda=1.5$',       f'{DATA}/exp_lam1.50/solution.npz'),
    ('hill, $\phi_i=0.3$',        f'{DATA}/hilltop_k1_phi0.300/solution.npz'),
    ('hill, $\phi_i=0.5$',        f'{DATA}/hilltop_k1_phi0.500/solution.npz'),
]
sols = [(label, load_sol(p)) for label, p in reps]
```

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
colors = plt.cm.viridis(np.linspace(0, 0.9, len(sols)))

for (label, s), c in zip(sols, colors):
    z = np.exp(-s['N']) - 1
    axes[0].plot(z, s['w_Q'], color=c, label=label)
    axes[1].plot(z, s['Omega_Q'], color=c, label=label)
    axes[2].plot(z, s['H_Q'], color=c, label=label)

axes[0].axhline(-1, ls=':', color='gray', alpha=0.5)
axes[0].set_xlabel('$z$'); axes[0].set_ylabel('$w_Q(z)$')
axes[0].set_title('Equation of state')
axes[0].set_xlim(0, 4)
axes[0].legend(fontsize=9)

axes[1].set_xlabel('$z$'); axes[1].set_ylabel('$\Omega_Q(z)$')
axes[1].set_title('Dark energy fraction')
axes[1].set_xlim(0, 4)

axes[2].set_xlabel('$z$'); axes[2].set_ylabel('$H_Q(z) / H_Q(0)$')
axes[2].set_title('Hubble parameter')
axes[2].set_xlim(0, 4)

plt.tight_layout()
plt.savefig(f'{FIGS}/fig1_evolution.png', dpi=140, bbox_inches='tight')
plt.show()
```

## 2. Best-fit $(w_0, w_a)$ comparison

The effective fluid model has
$$\Omega_{\rm fit}' = 3w_{\rm fit}\Omega_{\rm fit}(\Omega_{\rm fit} - 1), \qquad
H_{\rm fit}' = -\tfrac{3}{2}H_{\rm fit}(1 + w_{\rm fit}\Omega_{\rm fit}),$$
with $w_{\rm fit}(N) = w_0 + w_a(1 - e^N) = w_0 + w_a\,z/(1+z)$.

The fit minimizes
$$E = \max_{0\le z<4}\left|\frac{H_{\rm fit}(z) - H_Q(z)}{H_Q(z)}\right|$$
over $\{w_0, w_a, \Omega_{\rm fit}(0), H_{\rm fit}(0)\}$.

A central finding (Fig. 4 of the paper): $H(z)$ matches the quintessence model to sub-percent accuracy even when $w_{\rm fit}(z)$ differs from $w_Q(z)$ at $O(1)$ near $z = 0$.

```python
# Reconstruct H_fit and w_fit for the most-displaced hilltop
from scipy.integrate import solve_ivp

hil = np.load(f'{FITS}/hilltop_scan.npz')
i_ref = np.argmax(hil['param'])  # phi_i=0.5
w0, wa = hil['w0'][i_ref], hil['wa'][i_ref]
Om0, H0 = hil['Om0'][i_ref], hil['H0'][i_ref]
print(f'Reference: phi_i={hil["param"][i_ref]:.3f}, w0={w0:.4f}, wa={wa:.4f}, E={hil["E"][i_ref]:.2e}')

ref_sol = load_sol(f'{DATA}/hilltop_k1_phi0.500/solution.npz')
H_Q = np.interp(N_FIT, ref_sol['N'], ref_sol['H_Q'])
w_Q = np.interp(N_FIT, ref_sol['N'], ref_sol['w_Q'])

def fit_rhs(N, y, w0, wa):
    Omf, Hf = y
    wf = w0 + wa*(1 - np.exp(N))
    return [3*wf*Omf*(Omf - 1), -1.5*Hf*(1 + wf*Omf)]

sol = solve_ivp(fit_rhs, (0.0, N_MIN), [Om0, H0], args=(w0, wa),
                method='DOP853', rtol=1e-10, atol=1e-12, t_eval=N_FIT[::-1])
H_fit = sol.y[1][::-1]
w_fit = w0 + wa*(1 - np.exp(N_FIT))
```

```python
fig, axes = plt.subplots(2, 1, figsize=(8, 7), sharex=True)
z = np.exp(-N_FIT) - 1

axes[0].semilogy(z, np.abs((H_fit - H_Q) / H_Q))
axes[0].set_ylabel(r'$|H_{\rm fit} - H_Q| / H_Q$')
axes[0].set_title(f'Fit residual: hilltop $\\phi_i=0.5$')
axes[0].set_ylim(1e-6, 1e-2)

axes[1].plot(z, w_Q, label='$w_Q(z)$', lw=2)
axes[1].plot(z, w_fit, '--', label='$w_{\\rm fit}(z)$', lw=2)
axes[1].axhline(-1, ls=':', color='gray', alpha=0.5)
axes[1].set_xlabel('$z$'); axes[1].set_ylabel('$w(z)$')
axes[1].legend()
axes[1].set_xlim(0, 4)

plt.tight_layout()
plt.savefig(f'{FIGS}/fig2_fit_quality.png', dpi=140, bbox_inches='tight')
plt.show()
```

## 3. Best-fit $(w_0, w_a)$ curves

Each one-parameter family of quintessence models maps to a curve in the $(w_0, w_a)$ plane. The exponential $\Lambda$CDM limit ($\lambda = 0$) and the hilltop $\Lambda$CDM limit ($\phi_i \to 0$) both pin the curve at $(-1, 0)$. As the displacement parameter grows, points move into the sector
$$w_0 + w_a < -1,$$
which naively suggests NEC violation in the CPL parameterization but in fact corresponds to NEC-satisfying scalar fields with $w_Q(z) \ge -1$ everywhere.

```python
exp = np.load(f'{FITS}/exp_scan.npz')
hil = np.load(f'{FITS}/hilltop_scan.npz')

fig, ax = plt.subplots(figsize=(7, 6))
ax.plot(exp['w0'], exp['wa'], 'o-', color='C1', label='exponential ($\\lambda$ scan)')
ax.plot(hil['w0'], hil['wa'], 's-', color='C0', label='hilltop $k^2=1$ ($\\phi_i$ scan)')
ax.scatter([-1], [0], marker='*', s=200, color='k', zorder=5, label='$\\Lambda$CDM')

# w_0 + w_a = -1 line
w0_line = np.linspace(-1.0, -0.5, 50)
ax.plot(w0_line, -1 - w0_line, 'k--', alpha=0.6, label='$w_0 + w_a = -1$')

ax.set_xlabel('$w_0$'); ax.set_ylabel('$w_a$')
ax.set_xlim(-1.05, -0.55); ax.set_ylim(-0.6, 0.05)
ax.legend()
ax.set_title('Best-fit $(w_0, w_a)$ for thawing quintessence')

plt.tight_layout()
plt.savefig(f'{FIGS}/fig3_w0wa_curves.png', dpi=140, bbox_inches='tight')
plt.show()
```

## 4. Reliability of the compression

For the hilltop family, the residual error $E_{\min}$ grows as $\phi_i$ increases and the field rolls more sharply near $z = 0$. The fitted $w_0$ also drifts away from the true $w_Q(0)$ : the maximum-$H$ fit prioritizes matching the integrated expansion history, not the instantaneous equation of state.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

axes[0].loglog(hil['param'], hil['E'], 's-', color='C0')
axes[0].set_xlabel('$\\phi_i$'); axes[0].set_ylabel('$E_{\\min}$')
axes[0].set_title('Best-fit error vs initial displacement')

axes[1].plot(hil['param'], hil['wQ_0'], 'o-', label='$w_Q(0)$', color='C2')
axes[1].plot(hil['param'], hil['w0'], 's--', label='best-fit $w_0$', color='C0')
axes[1].set_xlabel('$\\phi_i$'); axes[1].set_ylabel('$w$')
axes[1].set_xscale('log')
axes[1].legend()
axes[1].set_title('True $w_Q(0)$ vs fitted $w_0$')

plt.tight_layout()
plt.savefig(f'{FIGS}/fig4_reliability.png', dpi=140, bbox_inches='tight')
plt.show()
```

## 5. Degeneracy in the $(w_0, w_a)$ plane

Around any best-fit point, a long, narrow valley of $(w_0, w_a)$ combinations matches $H(z)$ to similar accuracy. The principal axis of this valley has slope $\Delta w_a/\Delta w_0 \approx -5$ for typical models (Fig. 1 of the paper reports $-4.8$).

We compute the slope from the eigendecomposition of the covariance of grid points satisfying
$$E(w_0, w_a) \le E_{\rm threshold}.$$

```python
hm = np.load(f'{DIAG}/heatmap.npz')
w0g, wag = hm['w0_grid'], hm['wa_grid']
E_grid = hm['E_grid']
W0, WA = np.meshgrid(w0g, wag, indexing='ij')

# Eigendecomposition of acceptable region (E < threshold)
threshold = 5e-3
mask = E_grid < threshold
pts = np.column_stack([W0[mask] - hm['w0_best'], WA[mask] - hm['wa_best']])
cov = np.cov(pts.T)
vals, vecs = np.linalg.eigh(cov)
v_major = vecs[:, -1]
slope = v_major[1] / v_major[0]
print(f'Acceptable region (E < {threshold}): {mask.sum()} grid points')
print(f'Eigenvalues: {vals}')
print(f'Major axis slope (dwa/dw0): {slope:.2f}')
print(f'Eccentricity ratio sqrt(lambda_max/lambda_min): {np.sqrt(vals[-1]/vals[0]):.2f}')
```

```python
fig, ax = plt.subplots(figsize=(8, 6))

logE = np.log10(E_grid)
im = ax.pcolormesh(W0, WA, logE, cmap='viridis_r', shading='nearest')
plt.colorbar(im, ax=ax, label='$\\log_{10}(E)$')

# Contours
levels = np.log10([1e-3, 3e-3, 1e-2])
cs = ax.contour(W0, WA, logE, levels=levels, colors='white', linewidths=1, alpha=0.7)
ax.clabel(cs, inline=True, fontsize=9, fmt={l: f'$10^{{{l:.1f}}}$' for l in levels})

# Best fit + degeneracy direction
ax.plot(hm['w0_best'], hm['wa_best'], 'r*', markersize=18, label='best fit')
t = np.linspace(-0.1, 0.1, 50)
ax.plot(hm['w0_best'] + t*v_major[0], hm['wa_best'] + t*v_major[1],
        'r--', lw=1.5, label=f'principal axis (slope = {slope:.2f})')

ax.set_xlabel('$w_0$'); ax.set_ylabel('$w_a$')
ax.set_title(f'Degeneracy heatmap: hilltop $\\phi_i=0.5$')
ax.legend(loc='upper right')
ax.grid(False)

plt.tight_layout()
plt.savefig(f'{FIGS}/fig5_heatmap.png', dpi=140, bbox_inches='tight')
plt.show()
```

## 6. Sensitivity to the fitting metric

Three objective functions:
$$E_H^\infty = \max_{z<4}\left|\frac{\Delta H}{H_Q}\right|, \qquad
E_H^{\rm RMS} = \sqrt{\langle(\Delta H/H_Q)^2\rangle}, \qquad
E_{D_M}^\infty = \max_{0<z<4}\left|\frac{\Delta D_M}{D_{M,Q}}\right|$$
where $D_M(z) = \int_0^z dz'/H(z')$.

The paper notes that switching metrics shifts the best-fit slope by $\lesssim 10\%$. For each best-fit point, we tabulate the three errors evaluated against the same reference quintessence model.

```python
ms = np.load(f'{DIAG}/metric_sensitivity.npz')

print(f'{"metric":<10} {"w0":>9} {"wa":>9} {"E_max":>10} {"E_rms":>10} {"E_DM":>10}')
print('-' * 60)
for name in ['max_H', 'rms_H', 'max_DM']:
    p = ms[f'{name}_params']
    print(f'{name:<10} {p[0]:>9.4f} {p[1]:>9.4f} '
          f'{ms[f"{name}_err_max"]:>10.2e} '
          f'{ms[f"{name}_err_rms"]:>10.2e} '
          f'{ms[f"{name}_err_DM"]:>10.2e}')

# Slope between the three points
points = np.array([[ms[f'{n}_params'][0], ms[f'{n}_params'][1]] for n in ['max_H', 'rms_H', 'max_DM']])
print(f'\nBest-fit positions:')
for n, p in zip(['max_H', 'rms_H', 'max_DM'], points):
    print(f'  {n}: ({p[0]:.4f}, {p[1]:.4f})')
spread_w0 = points[:,0].max() - points[:,0].min()
spread_wa = points[:,1].max() - points[:,1].min()
print(f'\nSpread: dw0 = {spread_w0:.4f}, dwa = {spread_wa:.4f}')
```

## 7. Numerical convergence

For fixed-step RK4, the global error in $H(z)$ over $z\in[0,4]$ should scale as $\epsilon_H \propto h^4$ for smooth solutions. The reference solution is computed with adaptive DOP853 at $\rm rtol = 10^{-12}$, $\rm atol = 10^{-14}$.

For adaptive DOP853, the error tracks the requested tolerance.

```python
rk4_files = sorted(glob.glob(f'{CONV}/rk4_h*.npz'))
adaptive_files = sorted(glob.glob(f'{CONV}/adaptive_rtol*.npz'))

rk4_h = np.array([float(np.load(f)['h']) for f in rk4_files])
rk4_err = np.array([float(np.load(f)['err_H']) for f in rk4_files])
rk4_runtime = np.array([float(np.load(f)['runtime']) for f in rk4_files])
order = np.argsort(rk4_h)
rk4_h, rk4_err, rk4_runtime = rk4_h[order], rk4_err[order], rk4_runtime[order]

ad_rtol = np.array([float(np.load(f)['rtol']) for f in adaptive_files])
ad_err = np.array([float(np.load(f)['err_H']) for f in adaptive_files])
ad_nfev = np.array([int(np.load(f)['nfev']) for f in adaptive_files])
order = np.argsort(ad_rtol)[::-1]
ad_rtol, ad_err, ad_nfev = ad_rtol[order], ad_err[order], ad_nfev[order]

# Fit observed RK4 order
log_h = np.log(rk4_h)
log_e = np.log(rk4_err + 1e-30)
slope, _ = np.polyfit(log_h, log_e, 1)
print(f'Observed RK4 order: {slope:.2f} (theoretical: 4)')
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

# RK4 convergence
axes[0].loglog(rk4_h, rk4_err, 'o-', color='C0', label='RK4 (measured)')
ref_e = rk4_err[0] * (rk4_h / rk4_h[0])**4
axes[0].loglog(rk4_h, ref_e, ':', color='gray', label='$\\propto h^4$')
axes[0].set_xlabel('$h$ (step size in $s$)')
axes[0].set_ylabel('$\\max |H_{\\rm RK4} - H_{\\rm ref}|/H_{\\rm ref}$')
axes[0].set_title(f'Fixed-step RK4 convergence (slope = {slope:.2f})')
axes[0].legend()

# Adaptive
ad_err_safe = np.maximum(ad_err, 1e-16)
axes[1].loglog(ad_rtol, ad_err_safe, 's-', color='C2', label='DOP853 (measured)')
axes[1].loglog(ad_rtol, ad_rtol, ':', color='gray', label='$\\epsilon = $ rtol')
axes[1].set_xlabel('rtol')
axes[1].set_ylabel('max relative $H$ error')
axes[1].set_title('Adaptive DOP853 vs tolerance')
axes[1].legend()
axes[1].invert_xaxis()

plt.tight_layout()
plt.savefig(f'{FIGS}/fig6_convergence.png', dpi=140, bbox_inches='tight')
plt.show()
```

## Summary

1. **Quintessence dynamics** : Both potential families thaw away from $\Lambda$CDM at low $z$, with $w_Q$ rising above $-1$ as the field accelerates downhill. $\Omega_Q$ grows from $10^{-6}$ at high $z$ to the fiducial $0.7$ today, and $H_Q(z)/H_Q(0)$ exceeds unity at $z>0$ as expected.

2. **$(w_0, w_a)$ fits** : All thawing models map to the sector $w_0 > -1$, $w_0 + w_a < -1$, despite $w_Q(z) \ge -1$ everywhere. The maximum $H$ residual is sub-percent for every model in our scan.

3. **Compression reliability** : The fit error $E_{\min}$ grows monotonically with $\phi_i$ in the hilltop family, and the best-fit $w_0$ drifts visibly away from $w_Q(0)$ as the field rolls more sharply. The two-parameter compression is a faithful summary of $H(z)$ but not of $w_Q(z)$.

4. **Degeneracy** : The acceptable region in the $(w_0, w_a)$ plane is highly elongated, with principal-axis slope close to the paper's $-4.8$. This degeneracy is intrinsic to the parameterization and explains the orientation of observational likelihood contours.

5. **Metric sensitivity** : Switching from max-$H$ to RMS-$H$ or max-$D_M$ shifts the best-fit point by $\lesssim 10\%$ along the degeneracy axis, consistent with the paper.

6. **Numerical convergence** : RK4 shows the expected $O(h^p)$ scaling with $p$ near the theoretical value of $4$. Adaptive DOP853 tracks the requested tolerance.
